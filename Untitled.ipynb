```python

class Population:
    def __init__(self, init_members=[], fitness_popsize=10, diversity_popsize=10, max_samples=100, max_generations=100):
        self.members: List[Solution] = init_members
        self.fitness_popsize = fitness_popsize
        self.diversity_popsize = diversity_popsize
        self.popsize = self.fitness_popsize + self.diversity_popsize
        self.lock = Lock()
        self.max_samples, self.max_generations = max_samples, max_generations
        self.n_samples, self.n_generations = len(self.members), 0

        if len(self.members) > self.popsize:
            raise Exception("cannot initialize Population with larger member set than `popsize`")

    

    def apply_generational_dropout(self):
        
        # choose fitness survivors based on quota
        # choose diversity survivors based on quota, marginal of fitness
    
        with self.lock:
            self.members = sorted(self.members, key=lambda f: f.evaluation.fitness, reverse=False)[:self.popsize]

    def new_generation(self):
        self.n_generations += 1
        return self
    
    def register(self, solns):
        with self.lock:
            for soln in solns:
                self.members.append(soln)
                self.n_samples += 1

                ## I dont think I like this mechanism
                # if len(self.members) > self.popsize:
                #     self.survival_filter()
                #     self.n_generations += 1

    @property
    def is_completed(self):
        with self.lock:
            return self.n_samples > self.max_samples or self.n_generations > self.max_generations
```
