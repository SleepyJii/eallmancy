```python
import subprocess

def apply_git_diff(repo_path, diff_str):
    try:
        # Run `git apply` with input from the diff string
        subprocess.run(
            ['git', 'apply', '--whitespace=nowarn'],
            input=diff_str,
            text=True,
            cwd=repo_path,
            check=True
        )
        print("Diff applied successfully.")
    except subprocess.CalledProcessError as e:
        print("Failed to apply diff:", e)

# Example usage
diff_string = """
diff --git a/example.txt b/example.txt
index e69de29..b6fc4c6 100644
--- a/example.txt
+++ b/example.txt
@@ -0,0 +1,2 @@
+Hello
+World
"""
repo_dir = "/path/to/your/repo"

apply_git_diff(repo_dir, diff_string)
```
