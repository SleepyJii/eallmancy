```python
_nodes = lineage_graph.nodes(data='artifact')
_node_ids = [a for (a,b) in _nodes]


def distance_act(d):
    if math.isinf(d):
        return 1
    C = 10
    return min(max(1-C/d, 0), 1)

_vals = list(map(lambda ts: ts[1].fitness, _nodes))
def percentage_greater_than(query):
    return sum(query > v for v in _vals) / len(_vals)
def fitness_act(f):
    # TODO implement more carefully
    return percentage_greater_than(f)

from experiments.scievo.operations.utils import heritage_distances
distances = heritage_distances(lineage_graph)


fitnesses = dict()
for nidx in _node_ids:
    fitnesses[nidx] = fitness_act(_nodes[nidx].fitness)

from collections import defaultdict
cluster_memo = defaultdict(lambda : 0)

cluster_mems = []
def append_remove_elem(nidx):
    cluster_mems.append(nidx)
    for midx in set(_node_ids).difference(cluster_mems):
        ky = (max(midx,nidx), min(midx,nidx))
        cluster_memo[midx] += fitness_act(fitnesses[nidx]) * distance_act(distances[ky])
        del distances[ky]
    del fitnesses[nidx]
    if nidx in cluster_memo.keys():
        del cluster_memo[nidx]

k=4
for cidx in range(2,k+1):
    if cidx == 2:
        npairs = len(distances)
        probs = np.array(list(map(distance_act, distances.values())))
        probs = probs.astype(float) / probs.sum()
        pair_idx = np.random.choice(list(range(npairs)), p=probs)
        nidx, midx = list(distances.keys())[pair_idx]
        append_remove_elem(nidx)
        append_remove_elem(midx)
    else:
        probs = np.array(list(map( lambda ts: 1.0* ts[1] * fitness_act(fitnesses[ts[0]]), cluster_memo.items() )))
        probs = probs.astype(float) / probs.sum()
        nidx = np.random.choice(
            list(cluster_memo.keys()),
            p=probs
        )
        append_remove_elem(nidx)
```
